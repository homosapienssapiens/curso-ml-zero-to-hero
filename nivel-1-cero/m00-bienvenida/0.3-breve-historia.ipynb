## 1. 📖 La intuición: dos formas de aprender un idioma

Hay dos maneras de enseñarle español a alguien:
1. **Con un libro de gramática:** reglas, excepciones, excepciones a las excepciones… Funciona para frases sencillas, pero ningún libro cubre cómo habla la gente de verdad.

La historia de la IA es, en gran parte, el paso de la opción 1 a la opción 2:

```
1950 ─────────── 1990 ─────────── 2012 ─────────── hoy

  "el libro de       "aprender de      "aprender de      "leer casi todo
   gramática"         tablas"           datos crudos"     internet"
```

Y no fue un camino recto: la IA ha vivido **"veranos"** (entusiasmo, dinero, promesas) e **"inviernos"** (decepción, recortes de fondos).

## 3. 📖 La era de las reglas (1950–1990)

Los pioneros pensaban que la inteligencia era **lógica**: si escribimos suficientes reglas, la máquina será inteligente.

- **ELIZA (1966)** simulaba a una psicoterapeuta con reglas como *"si el usuario dice 'estoy ___', responde '¿por qué estás ___?'"*. Mucha gente creyó que la entendía.
- **Los sistemas expertos (años 80)** capturaban el conocimiento de médicos o ingenieros en miles de reglas "si… entonces…". Algunos funcionaron muy bien en tareas acotadas.

**¿Por qué no bastó?** Porque el mundo real tiene demasiados casos. Cada regla nueva choca con otras, mantenerlas es carísimo y el sistema se rompe ante lo imprevisto. A esto se le llamó el **cuello de botella del conocimiento**.

## 4. 📖 El perceptrón y el primer invierno

En 1958 Frank Rosenblatt construyó el **perceptrón**: una "neurona artificial" que **aprende** a partir de ejemplos ajustando unos números (pesos). La prensa de la época prometió máquinas que pronto caminarían, hablarían y serían conscientes.

En 1969 Marvin Minsky y Seymour Papert demostraron un límite incómodo: un perceptrón solo puede separar las clases con **una línea recta**. Hay problemas sencillísimos que no puede resolver, como el **XOR** ("o uno, o el otro, pero no los dos"):

| A | B | AND (los dos) | XOR (solo uno) |
|---|---|---|---|
| 0 | 0 | 0 | 0 |
| 0 | 1 | 0 | 1 |
| 1 | 0 | 0 | 1 |
| 1 | 1 | 1 | 0 |

Vamos a reproducir ese momento histórico.

````python
entradas = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
AND = [0, 0, 0, 1]
XOR = [0, 1, 1, 0]

def probar(modelo, salida_correcta, nombre):
    modelo.fit(entradas, salida_correcta)
    print(f"{nombre:<4} correcto: {salida_correcta}  predicho: {modelo.predict(entradas).tolist()}  "
          f"→ acierto {modelo.score(entradas, salida_correcta):.0%}")
````

### 🔮 Ejercicio 4 — ¿Podrá el perceptrón?
Vamos a entrenar un perceptrón con AND y otro con XOR. ¿Qué acierto esperas en cada caso?

✍️ *Mis predicciones:* AND → ___ % · XOR → ___ %

````python
probar(Perceptron(random_state=0), AND, "AND")
probar(Perceptron(random_state=0), XOR, "XOR")
````

<details><summary><b>👀 Ver respuesta</b></summary>

**AND → 100 %. XOR → 50 %** (lo mismo que adivinar). Dibuja los 4 puntos en un plano: para AND, una recta separa el (1,1) del resto. Para XOR, los dos "1" están en esquinas opuestas: **ninguna recta** los separa.

Ese resultado, y las promesas incumplidas, ayudaron a enfriar el entusiasmo por las redes neuronales. La financiación cayó y llegó el primer **invierno de la IA** (aprox. 1974–1980).
</details>

### 📖 La solución: apilar neuronas en capas
La salida era añadir **capas ocultas**: neuronas entre la entrada y la salida, que combinan rectas para formar fronteras más complejas. El problema era **cómo entrenarlas**. La respuesta, la **retropropagación** (*backpropagation*), se popularizó en 1986 (Lección 9.4).

### 🔮 Ejercicio 5
Ahora una red con **una capa oculta de 4 neuronas**. ¿Resolverá el XOR?

✍️ *Mi predicción:*

````python
red = MLPClassifier(hidden_layer_sizes=(4,), activation="tanh", solver="lbfgs", max_iter=1000, random_state=0)
probar(red, XOR, "XOR")
````

<details><summary><b>👀 Ver respuesta</b></summary>

**Sí: 100 %.** Con una capa oculta, la red combina varias rectas y "encierra" las dos esquinas. La misma idea, escalada a muchísimas capas y millones de neuronas, es el **Deep Learning**.

Pero en los años 80 y 90 las redes seguían siendo difíciles de entrenar: pocas computadoras potentes y pocos datos. Tras la caída de los sistemas expertos llegó un segundo invierno (aprox. 1987–1993).
</details>

## 6. 📖 2012: la revolución del Deep Learning

En el concurso **ImageNet** los equipos competían por clasificar fotos en 1,000 categorías. Una métrica clave era el **error top-5** (el porcentaje de fotos en que la respuesta correcta no estaba entre las 5 primeras apuestas del modelo).

````python
imagenet = pd.DataFrame({
    "año":    [2010, 2011, 2012, 2013, 2014, 2015],
    "error":  [28.2, 25.8, 16.4, 11.7, 6.7, 3.6],     # error top-5 del ganador (%)
    "modelo": ["ML clásico", "ML clásico", "AlexNet (DL)", "ZFNet (DL)", "GoogLeNet (DL)", "ResNet (DL)"],
})
ERROR_HUMANO = 5.1   # estimación de una persona entrenada

plt.bar(imagenet.año, imagenet.error, color=["gray", "gray", "C3", "C0", "C0", "C0"])
plt.axhline(ERROR_HUMANO, color="k", ls="--", label=f"Persona entrenada (~{ERROR_HUMANO} %)")
for a, e in zip(imagenet.año, imagenet.error):
    plt.text(a, e + 0.5, f"{e}", ha="center")
plt.ylabel("Error top-5 (%)"); plt.title("ImageNet: el año en que todo cambió"); plt.legend(); plt.show()
````

## 7. 📖 Transformers, LLMs e IA generativa (2017–hoy)

En 2017 un artículo de Google, *"Attention Is All You Need"*, presentó el **Transformer**: una arquitectura que lee todas las palabras de un texto a la vez y decide a cuáles "prestar atención" (Módulo 12). Escalaba de maravilla con más datos y más GPUs.

Así nacieron los **LLMs** (*Large Language Models*, grandes modelos de lenguaje). Mira cómo creció su tamaño:

````python
gpts = pd.DataFrame({
    "modelo":     ["GPT-1", "GPT-2", "GPT-3"],
    "año":        [2018, 2019, 2020],
    "parametros": [117e6, 1.5e9, 175e9],
})
plt.bar(gpts.modelo, gpts.parametros)
plt.yscale("log"); plt.ylabel("Parámetros (escala logarítmica)")
plt.title("Cada versión, ~10 a 100 veces más grande"); plt.show()

print(f"GPT-3 tiene {gpts.parametros[2] / gpts.parametros[0]:,.0f} veces más parámetros que GPT-1.")
````

**Paso 3 🔮 — ¿Cuánto tardaron las ideas?** Antes de ejecutar, estima cuántos años pasaron entre el perceptrón (1958) y AlexNet (2012).

````python
def años_entre(desde, hasta):
    return hasta - desde

print("Perceptrón → retropropagación popular:", años_entre(1958, 1986), "años")
print("Retropropagación → AlexNet:          ", años_entre(1986, 2012), "años")
print("Transformer → ChatGPT:               ", años_entre(2017, 2022), "años")
````

## 1. 📖 ¿Qué es un notebook?

Un **notebook** (cuaderno) es un documento que mezcla dos tipos de **celdas**:
- **Celdas de texto** (como esta): explicaciones, tablas, imágenes.
- **Celdas de código** (con fondo gris): instrucciones de Python que puedes **ejecutar**.

Detrás del notebook hay un **kernel**: el "cerebro" que ejecuta el código y **recuerda** lo que ya ejecutaste. Piensa en una calculadora con memoria: si guardas un número en la memoria, puedes usarlo después… pero si la apagas (reinicias el kernel), se borra todo.

| Acción | Cómo |

| Ejecutar todo el notebook | Menú *Ejecutar todo* / *Run all* (Colab: *Entorno de ejecución → Ejecutar todas*) |
| Borrar la memoria y empezar de cero | *Reiniciar* / *Restart* (Colab: *Entorno de ejecución → Reiniciar sesión*) |
| Editar una celda de texto | Doble clic sobre ella |

### Pieza 6 — Decisiones con `if`
Fíjate en los **dos puntos** y en la **sangría** (los espacios al inicio): así sabe Python qué líneas van dentro del `if`.

````python
temperatura = 28   # ✏️

if temperatura > 25:
    print("Hace calor ☀️")
elif temperatura > 15:
    print("Está agradable 🌤️")
else:
    print("Hace frío 🧥")
````

### Pieza 8 — Funciones: recetas reutilizables
Una función empaqueta unas instrucciones con un nombre. Recibe datos (entre paréntesis) y devuelve un resultado (`return`).

````python
def precio_con_iva(precio, iva=0.16):    # iva vale 0.16 si no le dices otra cosa
    return round(precio * (1 + iva), 2)    # redondeamos a 2 decimales

print(precio_con_iva(100))
print(precio_con_iva(100, iva=0.21))
````

**1,015 cafés** en la semana; el mejor día es el **sábado** (210).

**Paso 2 🔮 — ¿La lluvia ayuda o perjudica?** Antes de ejecutar, mira la tabla: ¿se venden más cafés los días con lluvia o sin lluvia?

✍️ *Mi predicción:*

````python
print(cafeteria.groupby("lluvia")["cafes"].mean())   # promedio de cafés con lluvia (1) y sin lluvia (0)
````

<details><summary><b>👀 Ver respuesta</b></summary>

Sin lluvia: **142** cafés de media · con lluvia: **152.5**. Parece que con lluvia se vende algo más… pero ¡cuidado! Solo hubo **2** días con lluvia y uno fue sábado, el mejor día de la semana. Con tan pocos datos no se puede concluir nada. Esta prudencia es el corazón del ML: **pocos datos → conclusiones frágiles** (Módulo 2).
</details>

**Paso 3 ✏️ — Una función útil.** Completa la función que estima cuántos cafés preparar: el promedio de la semana más un **margen** de seguridad.

````python
def cafes_a_preparar(tabla, margen=0.10):
    promedio = tabla["cafes"].mean()
    return round(promedio * (1 + margen))      # ✏️ prueba otros márgenes al llamarla

print(cafes_a_preparar(cafeteria))
print(cafes_a_preparar(cafeteria, margen=0.25))
````

<details><summary><b>👀 Ver respuesta</b></summary>

Con 10 % de margen: **160** cafés; con 25 %: **181**. (El promedio de la semana es exactamente 145.)

Fíjate: acabas de construir tu primer **"modelo"**, aunque sea muy simple: una regla que, a partir de datos del pasado, predice cuántos cafés preparar. Usar el promedio como predicción tiene nombre: es una **línea base** (Lección 4.7), y cualquier modelo de ML que hagas tendrá que ganarle.
</details>

**Paso 4 ▶️ — La gráfica del informe.** Ejecuta y luego responde: ¿qué días quedan **por encima** de la línea del promedio?

````python
plt.bar(cafeteria["dia"], cafeteria["cafes"], color=["steelblue" if l == 0 else "gray" for l in cafeteria["lluvia"]])
plt.axhline(cafeteria["cafes"].mean(), color="red", ls="--", label="promedio")
plt.title("Cafés por día (gris = llovió)"); plt.legend(); plt.show()
````

**Viernes, sábado y domingo** (160, 210 y 190) quedan por encima de 145. El patrón más claro no es la lluvia sino el **fin de semana**. Si fueras a construir un modelo, "¿es fin de semana?" sería una característica muy útil (Lección 2.4).

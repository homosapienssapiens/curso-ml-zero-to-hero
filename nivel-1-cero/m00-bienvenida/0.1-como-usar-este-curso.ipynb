## 1. 📖 La intuición: nadie aprende a andar en bicicleta leyendo

Puedes leer diez libros sobre cómo andar en bicicleta y caerte en el primer intento. Con el Machine Learning (ML) pasa igual: entenderlo de verdad requiere **hacer**, equivocarte y corregir.

Por eso este curso funciona como un **gimnasio**, no como una biblioteca:

## 4. ▶️ Una probadita: tu primer modelo en 30 segundos

Todavía no sabes ML, ¡y no importa! Vamos a ver uno en acción para que sepas hacia dónde vamos.

Tenemos **1,797 imágenes de dígitos escritos a mano** (del 0 al 9), de 8 × 8 píxeles. Mira algunas:

````python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

digitos = load_digits()
fig, axes = plt.subplots(1, 8, figsize=(10, 2))
for ax, imagen, etiqueta in zip(axes, digitos.images, digitos.target):
    ax.imshow(imagen, cmap="gray_r")
    ax.set_title(f"Es un {etiqueta}")
    ax.axis("off")
plt.show()
````

Le vamos a dar a un modelo **la mayoría de las imágenes junto con la respuesta correcta** para que aprenda, y guardaremos **450 imágenes que nunca verá** para examinarlo.

### 🔮 Ejercicio 5 — ¿Qué tan bien le irá?
Nadie le va a programar reglas como "un 0 es un círculo". Solo verá ejemplos. ¿Qué porcentaje de las 450 imágenes nuevas crees que acertará?
- a) Alrededor de 10 % (lo mismo que adivinar al azar)
- b) Alrededor de 50 %
- c) Alrededor de 80 %
- d) Más de 95 %

✍️ *Mi apuesta:*

````python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Separamos: imágenes para aprender y 450 imágenes "de examen" que el modelo no verá
X_aprender, X_examen, y_aprender, y_examen = train_test_split(
    digitos.data, digitos.target, test_size=450, random_state=0)

modelo = KNeighborsClassifier()           # un modelo sencillo (lo estudiarás en la Lección 5.3)
modelo.fit(X_aprender, y_aprender)        # aprende a partir de los ejemplos
acierto = modelo.score(X_examen, y_examen)  # lo examinamos con imágenes nuevas

print(f"Aprendió con {len(X_aprender)} imágenes y acertó el {acierto:.0%} de las {len(X_examen)} de examen.")
````

<details><summary><b>👀 Ver respuesta</b></summary>

**d) Acertó el 98 %.** Con tres líneas importantes (`KNeighborsClassifier()`, `fit` y `score`) y sin una sola regla escrita a mano. Eso es el Machine Learning: **aprender a partir de ejemplos** (lo verás a fondo en la Lección 1.1).

No te preocupes si el código te parece chino. Al final del Módulo 1 sabrás explicar cada línea.
</details>

### ✏️ Ejercicio 6 — ¿Y con menos ejemplos?
¿Qué pasa si el modelo aprende con muy **pocas** imágenes? Cambia `CUANTOS` a 10, 50 y 200 y vuelve a ejecutar cada vez. Antes de cada ejecución, 🔮 predice el porcentaje.

````python
CUANTOS = 10   # ✏️ prueba 10, 50 y 200

modelo_chico = KNeighborsClassifier().fit(X_aprender[:CUANTOS], y_aprender[:CUANTOS])
print(f"Con {CUANTOS} imágenes para aprender: acierto en el examen = {modelo_chico.score(X_examen, y_examen):.0%}")
````

✍️ *Mis resultados:* 10 → ___ · 50 → ___ · 200 → ___

<details><summary><b>👀 Ver respuesta</b></summary>

**10 → 22 %**, **50 → 74 %**, **200 → 94 %** (y 98 % con las 1,347 completas). Con solo 10 imágenes (¡una por dígito, en promedio!) apenas supera el azar; con 200 ya es muy bueno. Más ejemplos → mejor aprendizaje, pero cada vez se gana menos. Ya intuyes una gran verdad del ML: **los datos son el combustible** (Módulo 2).
</details>

### 8.3 🧩 Práctica integradora: "¿Cuántos ejemplos necesita?"

Vas a investigar, como un científico, cuántas imágenes necesita el modelo de dígitos para acertar al menos el **90 %**. Hazlo en 3 pasos cortos y revisa cada respuesta antes de seguir.

**Paso 1 🔮 — Tu hipótesis.** Con 1,347 imágenes acierta 98 %. ¿Cuántas crees que bastan para llegar al 90 %? ¿100? ¿300? ¿1,000?

✍️ *Mi hipótesis:*

**Paso 2 ▶️ — El experimento.** La celda entrena el modelo con distintas cantidades de imágenes y dibuja la curva.

````python
cantidades = [10, 20, 50, 100, 200, 400, 800, 1347]   # ✏️ puedes añadir valores
aciertos = []
for n in cantidades:
    m = KNeighborsClassifier().fit(X_aprender[:n], y_aprender[:n])
    aciertos.append(m.score(X_examen, y_examen))
    print(f"{n:>5} imágenes → {aciertos[-1]:.1%}")

plt.plot(cantidades, aciertos, "o-")
plt.axhline(0.9, color="gray", ls="--", label="meta: 90 %")
plt.xscale("log"); plt.xlabel("Imágenes para aprender (escala logarítmica)"); plt.ylabel("Acierto en el examen")
plt.legend(); plt.grid(alpha=.3); plt.title("Más ejemplos, mejor modelo… hasta cierto punto"); plt.show()
````

## 1. 📖 La intuición: muñecas rusas

Las cuatro palabras del título no son sinónimos ni rivales: son como **muñecas rusas** (matrioskas). Cada una vive dentro de la anterior.

```
┌─────────────────── INTELIGENCIA ARTIFICIAL (IA) ───────────────────┐
│  Cualquier técnica para que una máquina haga tareas "inteligentes" │
│  (incluye reglas escritas a mano, búsqueda, lógica…)               │
│   ┌──────────────── MACHINE LEARNING (ML) ──────────────────┐      │
│   │  La máquina APRENDE las reglas a partir de ejemplos     │      │
│   │   ┌────────────── DEEP LEARNING (DL) ──────────────┐    │      │
│   │   │  ML con redes neuronales de muchas capas       │    │      │
│   │   │   ┌──────── IA GENERATIVA ────────┐            │    │      │
│   │   │   │  CREA contenido nuevo:        │            │    │      │
│   │   │   │  texto, imágenes, audio…      │            │    │      │
│   │   │   └───────────────────────────────┘            │    │      │
│   │   └────────────────────────────────────────────────┘    │      │
│   └─────────────────────────────────────────────────────────┘      │
└────────────────────────────────────────────────────────────────────┘
```

| Término | Idea en una frase | Ejemplo |
|---|---|---|
| **IA** (*Artificial Intelligence*) | Máquinas que hacen tareas que parecen requerir inteligencia | El rival de ajedrez de un videojuego antiguo |
| **ML** (*Machine Learning*) | La máquina descubre las reglas **a partir de ejemplos** | Filtro de spam de tu correo |
| **DL** (*Deep Learning*) | ML con **redes neuronales profundas** (muchas capas) | Desbloqueo con reconocimiento facial |
| **IA generativa** (*Generative AI*) | Modelos que **crean** contenido nuevo | ChatGPT, generadores de imágenes |

> 📝 **Matiz:** casi toda la IA generativa actual usa Deep Learning, por eso la dibujamos dentro. Pero también existen técnicas generativas más simples, sin redes neuronales (verás una de juguete en la Sección 5).

## 2. 📖 IA clásica: las reglas las escribe una persona

La forma más antigua de IA: un experto piensa las reglas y un programador las escribe una por una. Por ejemplo, un asistente que te dice si llevar abrigo:

````python
def asistente_de_reglas(temperatura, llueve):
    # Reglas escritas A MANO por una persona
    if llueve:
        return "🧥 Lleva abrigo"
    if temperatura < 15:
        return "🧥 Lleva abrigo"
    return "👕 No hace falta"

print(asistente_de_reglas(temperatura=22, llueve=False))
````

- a) 10 °C, sin lluvia · b) 25 °C, con lluvia · c) 15 °C, sin lluvia

✍️ *Mis predicciones:* a → ___ · b → ___ · c → ___

````python
for temp, lluvia in [(10, False), (25, True), (15, False)]:
    print(f"{temp} °C, llueve={lluvia} →", asistente_de_reglas(temp, lluvia))
````

<details><summary><b>👀 Ver respuesta</b></summary>

El caso c) muestra el punto débil de las reglas a mano: **alguien tuvo que decidir que el límite es 15**. ¿Por qué no 16? Y si hay viento, ¿qué? Cada caso nuevo exige otra regla escrita por una persona.
</details>

## 3. 📖 Machine Learning: las reglas se aprenden de ejemplos

En vez de escribir las reglas, le damos a la máquina **ejemplos del pasado** con la respuesta correcta y le pedimos que **descubra las reglas sola**.

Supongamos que anotamos durante 200 días qué hizo Marta: si hacía frío, si llovía y si llevó abrigo. Marta nunca nos dijo su regla.

````python
rng = np.random.default_rng(0)
dias = pd.DataFrame({
    "temperatura": rng.integers(0, 35, 200),
    "llueve":      rng.integers(0, 2, 200),      # 1 = sí, 0 = no
})
# La costumbre real de Marta (el modelo NO la conoce):
dias["llevo_abrigo"] = ((dias.temperatura < 18) | (dias.llueve == 1)).astype(int)
dias.head()
````

### 🔮 Ejercicio 3 — ¿Descubrirá la regla de Marta?
Vamos a darle esos 200 días a un **árbol de decisión** (un modelo de ML que estudiarás en la Lección 5.4). Solo ve los ejemplos.

¿Crees que encontrará algo parecido a "abrigo si llueve **o** si hace menos de ___ grados"? ¿Qué límite de temperatura crees que encontrará?

✍️ *Mi apuesta:*

````python
modelo_marta = DecisionTreeClassifier(max_depth=2, random_state=0)
modelo_marta.fit(dias[["temperatura", "llueve"]], dias["llevo_abrigo"])   # ← aquí APRENDE

print(export_text(modelo_marta, feature_names=["temperatura", "llueve"], class_names=["sin abrigo", "abrigo"]))
````

<details><summary><b>👀 Ver respuesta</b></summary>

El árbol aprendió por sí solo: **"si llueve → abrigo; si no llueve, abrigo solo cuando la temperatura es ≤ 17.5"**. Es exactamente la regla de Marta (menos de 18 °C o lluvia), aunque haga las preguntas en otro orden. El límite 17.5 está justo entre 17 y 18: el modelo no sabe que Marta pensaba en "18", solo ve que con 17 llevó abrigo y con 18 no.

**Esta es la diferencia clave:** en la Sección 2 una persona escribió las reglas; aquí la máquina las **descubrió a partir de ejemplos**. Eso es Machine Learning.
</details>

## 4. 📖 Deep Learning: redes neuronales con muchas capas

El **Deep Learning** es un tipo de ML que usa **redes neuronales**: muchísimas "perillas" (números) organizadas en **capas**. Su superpoder es que aprende directamente de datos "crudos" (los píxeles de una foto, el sonido de una voz) sin que nadie le diga qué buscar.

```
 píxeles ──► [capa 1: trazos] ──► [capa 2: curvas, esquinas] ──► [capa 3: partes del dígito] ──► "es un 7"
```

Vamos a entrenar una red neuronal **pequeñita** para reconocer dígitos escritos a mano (los de la Lección 0.1).

````python
digitos = load_digits()
X_tr, X_te, y_tr, y_te = train_test_split(digitos.data, digitos.target, test_size=450, random_state=0)

red = MLPClassifier(hidden_layer_sizes=(32, 16), max_iter=1000, random_state=0)   # 2 capas ocultas
red.fit(X_tr, y_tr)
print(f"Acierto de la red neuronal con imágenes nuevas: {red.score(X_te, y_te):.1%}")
````

### 🔮 Ejercicio 5 — ¿Cuántas perillas?
Esta red tiene una capa de entrada (64 píxeles), dos capas ocultas (32 y 16 neuronas) y una de salida (10 dígitos). Cada conexión entre neuronas es una "perilla" que la red ajusta al aprender.

¿Cuántas perillas crees que tiene en total? a) unas 50 · b) unas 500 · c) unas 2,700 · d) unos 2 millones

✍️ *Mi apuesta:*

````python
perillas = sum(w.size for w in red.coefs_) + sum(b.size for b in red.intercepts_)
print(f"Esta red pequeñita tiene {perillas:,} perillas (parámetros).")
````

<details><summary><b>👀 Ver respuesta</b></summary>

**c) 2,778 perillas.** Y es una red de juguete: los modelos de lenguaje como los que hay detrás de ChatGPT tienen **miles de millones** de perillas. Por eso necesitan tantos datos y computadoras tan potentes.

> 📝 Aquí usamos `MLPClassifier` de scikit-learn porque es muy fácil de usar. Para redes grandes de verdad se usan librerías como **PyTorch** o **TensorFlow**; las verás desde el Módulo 9.
</details>

## 5. 📖 IA generativa: crear en lugar de clasificar

Hasta ahora los modelos **decidían** algo: "abrigo o no", "es un 7". La **IA generativa** hace otra cosa: **crea** contenido nuevo que se parece a lo que vio al entrenar.

| Pregunta discriminativa (decide) | Pregunta generativa (crea) |
|---|---|
| ¿Este correo es spam? | Escribe un correo para mi jefe |
| ¿Esta foto tiene un gato? | Dibuja un gato astronauta |
| ¿Esta reseña es positiva? | Escribe una reseña de este restaurante |

Construyamos una **IA generativa de juguete**: lee unos refranes, anota **qué palabra suele venir después de cada palabra**, y luego inventa refranes nuevos encadenando palabras. (No es Deep Learning: es la idea más simple posible. Pero los LLMs como ChatGPT hacen algo emparentado: predecir la siguiente palabra, con una red neuronal gigantesca. Lección 13.5.)

````python
refranes = '''
más vale pájaro en mano que cien volando
a quien madruga dios le ayuda
no por mucho madrugar amanece más temprano
camarón que se duerme se lo lleva la corriente
más vale tarde que nunca
en casa de herrero cuchillo de palo
dime con quién andas y te diré quién eres
el que mucho abarca poco aprieta
'''.strip().split("\n")

# "Entrenar": anotar qué palabras vienen después de cada palabra
siguientes = {}
for refran in refranes:
    palabras = ["<inicio>"] + refran.split() + ["<fin>"]
    for actual, sig in zip(palabras, palabras[1:]):
        siguientes.setdefault(actual, []).append(sig)

def inventar_refran(azar):
    palabra, frase = "<inicio>", []
    while True:
        palabra = azar.choice(siguientes[palabra])   # elige una de las palabras que vio después
        if palabra == "<fin>":
            return " ".join(frase)
        frase.append(palabra)

print("Después de 'más' vio:", siguientes["más"])
print("Después de 'que' vio:", siguientes["que"])
````

### 🔮 Ejercicio 6 — ¿Qué inventará?
Antes de ejecutar: ¿crees que el generador puede usar una palabra que **no aparece** en ningún refrán (por ejemplo, "gato")? ¿Por qué?

✍️ *Mi respuesta:*

````python
azar = random.Random(4)   # ✏️ cambia el 4 por otro número para obtener refranes distintos
for _ in range(6):
    print("🗣️", inventar_refran(azar))
````

<details><summary><b>👀 Ver respuesta</b></summary>

**No puede.** Solo elige entre palabras que vio en los datos de entrenamiento. Por eso sus "refranes" mezclan pedazos de los originales: algunos salen idénticos a uno real (*"más vale tarde que nunca"*), otros son combinaciones nuevas como *"camarón que mucho abarca poco aprieta"*, que suena a refrán pero nadie lo dijo nunca. Y como solo mira **una** palabra hacia atrás, a veces se enreda: *"…en casa de herrero cuchillo de herrero cuchillo de herrero…"*.

Ese es el corazón de la IA generativa: **producir algo nuevo que se parece a lo que vio**. Los LLMs hacen lo mismo con una escala enormemente mayor (millones de libros, miles de millones de perillas) y mirando muchas palabras hacia atrás, no solo una. Por eso escriben textos coherentes… y por eso a veces inventan datos falsos con total seguridad: las **alucinaciones** (Lección 13.9).
</details>

## 7. 🚨 Mitos y errores comunes

| 🚨 Mito | La realidad |
|---|---|
| "IA = ChatGPT" | ChatGPT es **una** IA generativa. La IA incluye desde un termostato con reglas hasta un recomendador de películas |
| "Todo ML es Deep Learning" | Para datos en tablas (ventas, clientes, créditos), los modelos clásicos y los árboles suelen **ganarle** al DL (Lección 6.6) |
| "Si es IA, es inteligente como una persona" | Cada modelo hace **una** tarea para la que fue entrenado. El modelo de Marta no sabe ajedrez |
| "La IA generativa entiende lo que dice" | Genera lo que es **probable** según sus datos de entrenamiento. Puede sonar seguro y estar equivocado |
| "Más complejo siempre es mejor" | Si una regla simple resuelve el problema, úsala (Lección 1.1) |

## 8. 🧠 Resumen en 5 puntos
1. **IA ⊃ ML ⊃ DL ⊃ (casi toda la) IA generativa**: como muñecas rusas.
2. **IA clásica:** una persona escribe las reglas. **ML:** la máquina las aprende de ejemplos.
3. **Deep Learning:** ML con redes neuronales de muchas capas y muchísimas perillas; brilla con datos crudos (imágenes, audio, texto).
4. **IA generativa:** crea contenido nuevo que se parece a lo que vio; no solo decide.
5. Ningún enfoque es "el mejor" siempre: depende del problema y de los datos.

### 9.3 🧩 Práctica integradora: "¿Riego hoy?"

Un jardinero lleva 300 días anotando la humedad de la tierra (%), si llovió el día anterior y si **regó** o no. Vas a comparar tus reglas contra las de un modelo de ML. Revisa cada respuesta antes de seguir.

**Paso 1 ▶️ — Mira los datos**

````python
rng2 = np.random.default_rng(1)
jardin = pd.DataFrame({
    "humedad":     rng2.integers(5, 80, 300),
    "llovio_ayer": rng2.integers(0, 2, 300),
})
# El criterio real del jardinero (secreto), con algunos días en que se equivocó:
regla_secreta = (jardin.humedad < 35) & (jardin.llovio_ayer == 0)
despistes = rng2.random(300) < 0.05            # 5 % de días hizo lo contrario
jardin["rego"] = (regla_secreta ^ despistes).astype(int)

print(f"Regó en {jardin.rego.mean():.0%} de los días")
jardin.head()
````

**Paso 2 ✏️ — Escribe tus reglas (IA clásica).** Mirando la tabla de arriba, completa la función con la regla que creas que usa el jardinero y ejecuta la celda para ver qué porcentaje de días aciertas.

````python
def mis_reglas(humedad, llovio_ayer):
    # ✏️ cambia la condición por la que creas que usa el jardinero
    if humedad < 50:
        return 1   # riega
    return 0       # no riega

predicciones = [mis_reglas(h, l) for h, l in zip(jardin.humedad, jardin.llovio_ayer)]
print(f"Mis reglas aciertan en el {np.mean(np.array(predicciones) == jardin.rego):.1%} de los días")
````

<details><summary><b>👀 Ver respuesta</b></summary>

La regla de partida (`humedad < 50`) acierta solo el **59.7 %**. Si pruebas `humedad < 35 and llovio_ayer == 0` llegarás al **96.7 %**: es la regla secreta, y el 5 % restante son los despistes del jardinero, que **ninguna** regla puede adivinar.

¿Cuánto tardaste en encontrarla? Imagina hacerlo con 50 columnas en lugar de 2.
</details>

**Paso 3 ▶️ — Ahora que aprenda la máquina (ML).** Un árbol de decisión recibe los mismos 300 días y busca las reglas solo.

````python
modelo_jardin = DecisionTreeClassifier(max_depth=2, random_state=0)
modelo_jardin.fit(jardin[["humedad", "llovio_ayer"]], jardin["rego"])

print(export_text(modelo_jardin, feature_names=["humedad", "llovio_ayer"], class_names=["no riega", "riega"]))
print(f"El modelo acierta en el {modelo_jardin.score(jardin[['humedad', 'llovio_ayer']], jardin['rego']):.1%} de los días")
````
